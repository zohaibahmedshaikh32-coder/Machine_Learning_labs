```python
import pandas as pd

# Load dataset directly from URL
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("=== TASK 1: DATASET CLEANING ===")

# 1. Display the first 10 rows
print("\n1. First 10 rows of the dataset:")
display(df.head(10))

# 2. Check for missing values in each column
print("\n2. Missing values count per column:")
print(df.isnull().sum())

# 3. Fill missing values in 'Age' column with the mean age
mean_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(mean_age)
print(f"\n3. Filled missing 'Age' values with mean age: {mean_age:.2f}")

# 4. Drop rows where 'Embarked' column is missing
df = df.dropna(subset=['Embarked'])
print(f"4. Remaining missing values in 'Embarked': {df['Embarked'].isnull().sum()}")

print("\nTask 1 Completed!")
```

```python
print("=== TASK 2: ENCODING CATEGORICAL DATA ===")

# 1. Convert 'Sex' column into numeric (0 = Male, 1 = Female)
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
print("\n1. 'Sex' column encoded (0 = Male, 1 = Female):")
display(df[['Sex']].head())

# 2. Apply One-Hot Encoding on 'Embarked' column
df = pd.get_dummies(df, columns=['Embarked'], prefix='Embarked', dtype=int)
print("\n2. One-Hot Encoded 'Embarked' columns added:")
display(df.filter(like='Embarked').head())

print("\nTask 2 Completed!")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

print("=== TASK 3: FEATURE SCALING & SPLITTING ===")

# 1. Select features: Age, Fare, Sex, Pclass
features = ['Age', 'Fare', 'Sex', 'Pclass']
X = df[features]
y = df['Survived']  # Target variable

print("\n1. Selected Features shape:", X.shape)

# 2. Apply StandardScaler to normalize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for clean interactive display
X_scaled_df = pd.DataFrame(X_scaled, columns=features)
print("\n2. First 5 rows of Scaled Features (Mean=0, Std=1):")
display(X_scaled_df.head())

# 3. Split data into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.20, random_state=42
)

print("\n3. Dataset Split Results:")
print(f"   - Training Set (X_train): {X_train.shape[0]} samples (80%)")
print(f"   - Testing Set  (X_test) : {X_test.shape[0]} samples (20%)")

print("\nTask 3 Completed!")
```
